```python
import time

li1 = [1, 2, 3, 4]
li2 = [5, 6, 7, 8]
start = time.time()
for i in range(len(li1)):
    li1[i] += li2[i]
print(li1)                     # [6, 8, 10, 12]
print(f"Time taken for the operation: {time.time() - start}")


import numpy as np
import time

a1 = np.array([1, 2, 3, 4, 9, 8, 6])
a2 = np.array([5, 6, 7, 8, 3, 5, 6])
start = time.time()
a3 = a1 + a2
print(a3)                     # [6, 8, 10, 12]
print(f"Time taken for the operation: {time.time() - start}")

```

# Combining Arrays

```python
# np.concatenate() --	Existing axis par arrays join karna
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.concatenate((a, b)))

a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print(np.concatenate((a, b), axis=0))

# np.stack()	     -- New axis par arrays combine karna
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.stack((a, b)))

# np.hstack()      --	Horizontal combination
print(np.hstack((a, b)))

# np.vstack()      --	Vertical combination
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.vstack((a,b)))


a = np.array([[1, 2],
              [3, 4]])

b = np.array([[5, 6],
              [7, 8]])

print(np.vstack((a,b)))

# vstack()	1D arrays ko 2D banakar rows mein rakhta hai
# hstack()	1D arrays ko side-by-side jodta hai

# np.column_stack()--	1D arrays ko columns mein arrange karna
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.column_stack((a, b)))
```
